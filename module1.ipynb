```python
# Student Marks Analyzer application

students = [
    {"name": "Rahul", "marks": 85},
    {"name": "Priya", "marks": 92},
    {"name": "Arjun", "marks": 68},
    {"name": "Sneha", "marks": 76},
    {"name": "Kiran", "marks": 95}
]

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

for student in students:
    grade = get_grade(student["marks"])
    print(
        student["name"],
        "- Marks:",
        student["marks"],
        "- Grade:",
        grade
    )
marks = []

for student in students:
    marks.append(student["marks"])

average = sum(marks) / len(marks)

print("Class Average:", average)
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
```
